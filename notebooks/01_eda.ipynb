This notebook performs data acquisition inspection, null value handling, feature distribution analysis, and text token profiling for both the TMDB 5000 Movies and IMDB 50k Reviews datasets.


```python
import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-darkgrid' if 'seaborn-v0_8-darkgrid' in plt.style.available else 'default')
print("Libraries imported successfully!")
```

### 1. Load Datasets
We load the raw datasets from `data/raw/`:
1. `tmdb_5000_movies.csv`: Movie metadata (overview, genres, keywords, ratings)
2. `imdb_reviews.csv`: User reviews labeled positive or negative


```python
movies_df = pd.read_csv("../data/raw/tmdb_5000_movies.csv")
reviews_df = pd.read_csv("../data/raw/imdb_reviews.csv")

print(f"Movies Dataset Shape: {movies_df.shape}")
print(f"Reviews Dataset Shape: {reviews_df.shape}")
```

### 2. Missing Value & Null Inspection


```python
print("--- Movies Missing Values ---")
print(movies_df[['title', 'genres', 'overview', 'keywords', 'vote_average', 'popularity']].isnull().sum())

print("\n--- Reviews Missing Values ---")
print(reviews_df.isnull().sum())
```

### 3. Movie Rating & Popularity Distributions


```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))

ax1.hist(movies_df['vote_average'].dropna(), bins=30, color='#f5c518', edgecolor='black')
ax1.set_title("Distribution of Movie Ratings (vote_average)")
ax1.set_xlabel("Rating (0 - 10)")
ax1.set_ylabel("Count")

ax2.hist(movies_df['popularity'].dropna(), bins=40, color='#e50914', edgecolor='black', log=True)
ax2.set_title("Distribution of Popularity (Log Scale)")
ax2.set_xlabel("Popularity Metric")
ax2.set_ylabel("Log Count")

plt.tight_layout()
plt.show()
```

### 4. Review Sentiment Distribution & Text Length Analysis


```python
reviews_df['review_length'] = reviews_df['review'].apply(lambda x: len(str(x).split()))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4))
reviews_df['sentiment'].value_counts().plot(kind='bar', ax=ax1, color=['#10b981', '#ef4444'], edgecolor='black')
ax1.set_title("IMDB Review Sentiment Class Balance")
ax1.set_xlabel("Sentiment")
ax1.set_ylabel("Count")

ax2.hist(reviews_df['review_length'], bins=40, color='#3b82f6', edgecolor='black')
ax2.set_title("Distribution of Word Counts in Reviews")
ax2.set_xlabel("Word Count")
ax2.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### 5. Summary of EDA Findings
- The IMDB dataset is balanced with 25,000 positive and 25,000 negative reviews.
- The average review length is approximately 230 words.
- In the movie catalog, ratings peak around 6.0 to 7.0 out of 10.
- Preprocessing steps required:
  - Fill missing overviews and extract genre strings from JSON structure.
  - HTML tag removal, regex sanitization, and lemmatization for review texts.
